## 4. Скалярное произведение · 55–65 минут


### Сценарий преподавателя

**Спросить.** «Как из нескольких признаков и весов получить одно число?»

**Не говорить сразу.** Не начинать с обозначения суммы или транспонирования.

**Ожидаемый ответ.** Перемножить соответствующие пары и сложить результаты.

**Ключевая мысль.** Скалярное произведение объединяет взвешенные признаки в одно число.

**Что показать.** Ручной пример `[1,2]·[3,4]`, затем `np.dot` и `@`.

**Практика.** Изменить один вес и предсказать направление изменения результата.

**Типичная ошибка.** Перемножать векторы покомпонентно, но забывать сложить результаты.

**Если не понимают.** Сделать две стопки пар «признак × вес», посчитать вклад каждой и сложить.

**Если осталось время.** Попросить придумать два разных набора, дающих один и тот же итог.

**Тайминг.** 10 минут


\[ [1,2] \cdot [3,4] = 1\cdot3 + 2\cdot4 = 11 \]


````python
x_small = np.array([1, 2])
w_small = np.array([3, 4])
print(np.dot(x_small, w_small))
print(x_small @ w_small)
````

Для ручного счёта используем условно нормализованные числа. Не вводите нормализацию как отдельную тему: достаточно сказать, что признаки заранее приведены к удобному масштабу.


````python
flight = np.array([0.8, 0.6, 0.4, 0.2])
weights = np.array([0.2, 0.4, 0.7, 0.3])
````

### До запуска

Считаем вместе: `0.16 + 0.24 + 0.28 + 0.06 = 0.74`.


````python
score = flight @ weights
score
````

### Проверка ручного вычисления


````python
manual_score = 0.8*0.2 + 0.6*0.4 + 0.4*0.7 + 0.2*0.3
print(manual_score)
print('Совпало:', np.isclose(score, manual_score))
````

### Практика · решение


````python
changed_weights = weights.copy()
changed_weights[2] = 0.0
print('Было:', score)
print('Стало:', flight @ changed_weights)
````

## 9. Финал · 88–90 минут


### Сценарий преподавателя

**Спросить.** «Кто объяснит `X @ w + b` без чтения формулы по буквам?»

**Не говорить сразу.** Не исправлять формулировку ученика на полуслове; сначала дать закончить всю цепочку.

**Ожидаемый ответ.** `X` — объекты и признаки, `w` — веса, `b` — смещение, результат — prediction для каждой строки.

**Ключевая мысль.** Ученик понимает смысл вычисления, а не только синтаксис NumPy.

**Что показать.** Финальную вертикальную схему.

**Практика.** Exit ticket: объяснить формулу соседу за 30 секунд.

**Типичная ошибка.** Повторять определения вместо объяснения процесса.

**Если не понимают.** Вернуться к одному объекту и пройти путь пальцем: четыре числа → четыре вклада → сумма → смещение.

**Если осталось время.** Попросить связать сегодняшнюю схему с будущей Computer Vision моделью.

**Тайминг.** 2 минуты


```text
REAL WORLD OBJECT
↓
FEATURES
↓
VECTOR x
↓
MANY OBJECTS
↓
MATRIX X
↓
WEIGHTS w
↓
X @ w + b
↓
PREDICTION
```

### Ожидаемая финальная формулировка ученика

`X` хранит объекты по строкам и признаки по столбцам. У каждого признака есть соответствующий вес в `w`. Операция `X @ w` считает взвешенную сумму для каждой строки, `b` сдвигает результат, и получается по одному prediction на объект.

### Мост к следующей паре

«В Computer Vision исходным объектом станет изображение. Готовая модель сама построит большое числовое представление и выполнит множество операций с обученными весами. Мы не увидим каждую операцию, но уже понимаем базовую идею внутри».
