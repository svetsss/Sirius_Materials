# Как модель видит данные

Мастер-класс для 9–10 классов · 90 минут

Сегодня мы разберём одну цепочку:

```text
реальный объект → признаки → вектор → матрица X → веса w
→ скалярное произведение → X @ w + b → prediction
```

> Это учебный авиационный пример, а не реальная модель управления полётами.

## 4. Скалярное произведение · 55–65 минут

### 📖 Маленький пример руками

\[
[1,2] \cdot [3,4] = 1\cdot3 + 2\cdot4 = 11
\]

Берём признаки → учитываем веса → складываем вклады → получаем одно число.


### ❓ До запуска

Совпадут ли результаты `np.dot(...)` и оператора `@`?


````python
x_small = np.array([1, 2])
w_small = np.array([3, 4])

print(np.dot(x_small, w_small))
print(x_small @ w_small)
````

Для удобного ручного счёта используем условно нормализованные числа от 0 до 1. Это отдельное числовое представление того же типа признаков, а не реальные авиационные значения.


````python
flight = np.array([0.8, 0.6, 0.4, 0.2])
weights = np.array([0.2, 0.4, 0.7, 0.3])
````

### ❓ До запуска

Посчитайте вручную: `0.8×0.2 + 0.6×0.4 + 0.4×0.7 + 0.2×0.3`.


````python
score = flight @ weights
score
````

`score` — условный результат линейного вычисления. Это **не** реальное время посадки и не решение для авиации.
